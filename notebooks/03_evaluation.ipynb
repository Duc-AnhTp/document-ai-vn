# 03 — So sánh Kết quả E1 vs E2 vs E3

Load metrics từ `results/` và vẽ biểu đồ so sánh.

```python
import json, os
import pandas as pd
import matplotlib.pyplot as plt

experiments = {
    'E1 (PaddleOCR)': 'results/e1_baseline/metrics.json',
    'E2 (Donut)': 'results/e2_donut/metrics.json',
    'E3 (SROIE)': 'results/e3_cross/metrics.json',
}

results = {}
for name, path in experiments.items():
    if os.path.exists(path):
        with open(path, 'r') as f:
            results[name] = json.load(f)
        print(f'{name}: F1={results[name]["overall"]["f1"]}')
    else:
        print(f'{name}: chua co ket qua')
```

```python
# Bang so sanh tong hop
rows = []
for name, m in results.items():
    rows.append({'Experiment': name, **m['overall']})
df = pd.DataFrame(rows)
print(df.to_string(index=False))
```

```python
# Bar chart F1 per field
fields = ['store_name', 'date', 'total', 'address']
fig, ax = plt.subplots(figsize=(10, 5))
x = range(len(fields))
width = 0.25
for i, (name, m) in enumerate(results.items()):
    f1s = [m['per_field'][f]['f1'] for f in fields]
    ax.bar([xi + i*width for xi in x], f1s, width, label=name)
ax.set_xticks([xi + width for xi in x])
ax.set_xticklabels(fields)
ax.set_ylabel('F1'); ax.set_title('F1 per Field'); ax.legend()
plt.tight_layout(); plt.show()
```

```python
# Bar chart overall metrics
metrics_names = ['f1', 'precision', 'recall']
fig, ax = plt.subplots(figsize=(8, 5))
x = range(len(metrics_names))
width = 0.25
for i, (name, m) in enumerate(results.items()):
    vals = [m['overall'][mn] for mn in metrics_names]
    ax.bar([xi + i*width for xi in x], vals, width, label=name)
ax.set_xticks([xi + width for xi in x])
ax.set_xticklabels(metrics_names)
ax.set_ylabel('Score'); ax.set_title('Overall Metrics'); ax.legend()
ax.set_ylim(0, 1.0)
plt.tight_layout(); plt.show()
```

```python
# Error analysis E3
ea_path = 'results/e3_cross/error_analysis.json'
if os.path.exists(ea_path):
    with open(ea_path, 'r') as f:
        ea = json.load(f)
    print('Error counts per field:', ea['error_counts'])
    print('\nVi du loi:')
    for ex in ea.get('examples', []):
        print(f"  [{ex['field']}] expected='{ex['expected']}' got='{ex['predicted']}'")
else:
    print('Chua co error analysis')
```

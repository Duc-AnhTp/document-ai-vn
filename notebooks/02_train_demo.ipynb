```python
import pandas as pd
import matplotlib.pyplot as plt

# Ve loss curve
log_path = 'results/e2_donut/training_log.csv'
try:
    df = pd.read_csv(log_path)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    ax1.plot(df['epoch'], df['train_loss'], label='Train')
    ax1.plot(df['epoch'], df['val_loss'], label='Val')
    ax1.set_xlabel('Epoch'); ax1.set_ylabel('Loss'); ax1.legend(); ax1.set_title('Loss Curve')
    ax2.plot(df['epoch'], df['val_f1'], 'g-o', label='F1')
    ax2.plot(df['epoch'], df['val_precision'], 'b--', label='Precision')
    ax2.plot(df['epoch'], df['val_recall'], 'r--', label='Recall')
    ax2.set_xlabel('Epoch'); ax2.set_ylabel('Score'); ax2.legend(); ax2.set_title('Validation Metrics')
    plt.tight_layout(); plt.show()
except FileNotFoundError:
    print(f'Chua co log: {log_path}')
```

# 03 - Donut evaluation

Notebook load `results/e2_donut/metrics.json` va visualize ket qua Donut tren MC-OCR test.

```python
import json
import os
import pandas as pd
import matplotlib.pyplot as plt

metrics_path = 'results/e2_donut/metrics.json'
if os.path.exists(metrics_path):
    with open(metrics_path, 'r', encoding='utf-8') as f:
        metrics = json.load(f)
    print('Overall:', metrics['overall'])
else:
    metrics = None
    print(f'No metrics found yet: {metrics_path}')
```

```python
if metrics:
    rows = [{'field': field, **values} for field, values in metrics['per_field'].items()]
    df = pd.DataFrame(rows)
    print(df.to_string(index=False))
```

```python
if metrics:
    fields = list(metrics['per_field'].keys())
    f1s = [metrics['per_field'][field]['f1'] for field in fields]
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.bar(fields, f1s)
    ax.set_ylim(0, 1)
    ax.set_ylabel('F1')
    ax.set_title('Donut F1 per field')
    plt.tight_layout(); plt.show()
```
